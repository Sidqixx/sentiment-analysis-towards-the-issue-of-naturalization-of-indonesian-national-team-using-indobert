```python
opinion_keywords = [
    # kata sikap eksplisit
    "menurut", "gue", "gw", "aku", "saya", "kami",
    "setuju", "dukung", "tolak", "parah", "bagus", "buruk",
    "hebat", "keren", "jelek", "bangga", "kecewa",

    # modalitas / saran
    "harus", "harusnya", "sebaiknya", "perlu", "wajib",

    # prediksi / spekulasi
    "bakal", "akan", "kayaknya", "sepertinya", "mungkin", "bisa jadi",

    # framing diskusi
    "menurutmu", "gimana", "kenapa", "apakah", "siapakah"
]

news_keywords = [
    "dpr", "komisi", "pssi", "resmi", "raker",
    "proses", "naturalisasi", "disetujui", "ditetapkan",
    "dilaporkan", "menyatakan", "ujar", "berita",
    "tanggal", "fifa matchday", "menjadi wni"
]

news_banget = [
    "resmi", "raker",
    "disetujui", "ditetapkan",
    "dilaporkan", "menyatakan",
    "berita timnas",
    "breaking news",
    "resmi diumumkan",
    "resmi mengumumkan",
    "resmi menetapkan",
    "dpr setujui",
    "dpr menyetujui",
    "komisi x",
    "rapat kerja",
    "ditetapkan sebagai",
    "telah disahkan",
    "secara resmi",
    "siaran pers",
    "press release",
    "dilantik",
    "menjadi wni",
    "mengucapkan sumpah",
    "resmi jadi",
    "diumumkan hari ini",
]
```

```python
def classify_type(text):
    text = str(text).lower()

    strong_news = any(w in text for w in news_banget )
    opinion_score = sum(1 for w in opinion_keywords if w in text)
    news_score    = sum(1 for w in news_keywords if w in text)
    has_question  = "?" in text

    if strong_news:
        return "information"
    elif opinion_score > 0 or has_question:
        return "opinion"
    elif news_score >= 2:
        return "information"
    else:
        return "opinion"
```

```python
import pandas as pd

df_filter = pd.read_csv("preprocessed_data.csv")
df_filter["type"] = df_filter["clean_text"].apply(classify_type)
```

```python
df_information = df_filter[df_filter["type"] == "information"]
df_information.sample(50)

df_information.to_csv("filter_information.csv", index=False)
```
