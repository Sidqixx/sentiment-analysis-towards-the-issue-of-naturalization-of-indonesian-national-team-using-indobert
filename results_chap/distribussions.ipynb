```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load Data
train_df = pd.read_csv("../data_labelling/train_labeled.csv")

# Cek isi asli datanya! Kalau ini kosong, berarti path CSV lu bermasalah
print("=== ISI KOLOM LABEL ASLI ===")
print(train_df['label'].value_counts())
print("============================\n")
```

```python
# 2. Hitung jumlah dan paksa urutkan dari 0, 1, 2
counts = train_df['label'].astype(int).value_counts().sort_index()

# Ganti nama index 0, 1, 2 jadi teks biar gampang dibaca
# Asumsi: 0=Negative, 1=Neutral, 2=Positive
counts.index = ['Negative', 'Neutral', 'Positive'] 
```

```python
# 3. Plotting super basic pake Pandas bawaan (Anti-Gagal)
plt.figure(figsize=(8, 6))
ax = counts.plot(kind='bar', color=['#ff6b6b', '#c8d6e5', '#1dd1a1'], edgecolor='black')

# Save dan munculin
plt.savefig('sentiment_distribution_train_FIX.png', dpi=300)
print("✅ Grafik berhasil disave! Cek file 'sentiment_distribution_train_FIX.png'")
```
