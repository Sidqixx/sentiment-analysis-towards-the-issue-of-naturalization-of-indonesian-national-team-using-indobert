```python
import pandas as pd

print("=== MEMULAI PENGGABUNGAN DATA FIXED ===")

# 1. Load semua file yang udah diselamatin temen lu
# (Pastiin path folder/nama file-nya udah bener ya)
train_df = pd.read_csv("../fixed/train_labeled_fixed.csv")
val_df = pd.read_csv("../fixed/val_labeled_fixed.csv")
test_df = pd.read_csv("../fixed/test_labeled_fixed.csv")
pool_df = pd.read_csv("../fixed/pseudo_pool_fixed.csv")

# 2. Gabungin semuanya jadi satu DataFrame raksasa
# Kita pake ignore_index=True biar index-nya ngurut rapi dari 0 lagi
full_df = pd.concat([train_df, val_df, test_df, pool_df], ignore_index=True)

print(f"✅ Total data setelah digabung: {len(full_df)} baris (Harusnya pas 2000 data)")

# ==========================================
# 3. MAGIC TRICK: Ubah teks createdAt jadi tipe data Waktu (DateTime)
# ==========================================
# Kadang Twitter ngasih zona waktu kayak "+0000", kita paksa parsing.
# parameter errors='coerce' buat ngubah data yang bener-bener ancur (kalo masih ada) jadi NaN
full_df['createdAt'] = pd.to_datetime(full_df['createdAt'], errors='coerce')

# Cek apakah ada tanggal yang gagal di-parsing
tanggal_rusak = full_df['createdAt'].isna().sum()
if tanggal_rusak > 0:
    print(f"⚠️ Warning: Ada {tanggal_rusak} baris yang tanggalnya masih error/kosong.")
    # Opsional: full_df = full_df.dropna(subset=['createdAt']) # Aktifin ini kalo mau buang yang error

# 4. Urutin data berdasarkan waktu (dari yang paling lama ke paling baru)
full_df = full_df.sort_values(by='createdAt').reset_index(drop=True)

# 5. Cek hasilnya bentar biar yakin
print("\n=== PREVIEW DATA (UDAH URUT WAKTU) ===")
print(full_df[['createdAt', 'label', 'cleaned_text']].head())

# 6. Save file gabungannya biar lu punya 1 Master File yang Suci!
full_df.to_csv("master_dataset_fixed.csv", index=False)
print("\n✅ MANTAP! File berhasil digabung dan disave jadi 'master_dataset_fixed.csv'!")
```
