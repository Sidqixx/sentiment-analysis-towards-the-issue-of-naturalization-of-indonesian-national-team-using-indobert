```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("=== GENERATE CONFUSION MATRIX IMAGES (SHORTCUT) ===")

# 1. Masukin angka mentah yang lu kasih ke gue kemaren
# Array Model 1 (Manual 1100)
cm_model1 = np.array([[100,  10,  20],
                      [ 15,  34,  11],
                      [  5,   3,  52]])

# Array Model 3 (CBST 1117)
cm_model3 = np.array([[106,  10,  14],
                      [ 15,  37,   8],
                      [  6,   4,  50]])

# Label kelas lu (Asumsi: 0=Negative, 1=Neutral, 2=Positive)
class_names = ['Negative', 'Neutral', 'Positive']

# 2. Fungsi buat ngegambar Matrix biar estetik
def plot_confusion_matrix(cm, title, filename):
    plt.figure(figsize=(6, 5))
    
    # Bikin heatmap pake seaborn (pake warna Biru/Blues biar kayak paper jurnal)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=class_names, yticklabels=class_names,
                annot_kws={"size": 14, "weight": "bold"})
    
    # Percantik teks
    plt.title(title, fontsize=14, fontweight='bold', pad=15)
    plt.ylabel('Actual Label (True)', fontsize=12, fontweight='bold')
    plt.xlabel('Predicted Label', fontsize=12, fontweight='bold')
    
    # Miringin label sumbu Y biar enak dibaca
    plt.yticks(rotation=0)
    
    plt.tight_layout()
    plt.savefig(filename, dpi=300)
    print(f"✅ Gambar '{filename}' berhasil disave!")
    plt.show()

# 3. Eksekusi gambarnya!
plot_confusion_matrix(cm_model1, 'Confusion Matrix: Model 1 (HAB)', 'cm_model1_hab.png')
plot_confusion_matrix(cm_model3, 'Confusion Matrix: Model 3 (CBST)', 'cm_model3_cbst.png')
```
